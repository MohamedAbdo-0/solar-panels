```python
import pandas as pd
import matplotlib .pyplot as plt

d1433 =(pd .read_csv ("system_1433.csv",parse_dates =["timestamp_local"])
.sort_values ("timestamp_local").reset_index (drop =True ))
d10 =(pd .read_csv ("system_10.csv",parse_dates =["timestamp_local"])
.sort_values ("timestamp_local").reset_index (drop =True ))

def missing_runs (df ,col ):
    miss =df [col ].isna ()
    start =miss &~miss .shift (fill_value =False )
    gid =start .cumsum ()
    runs =(df [miss ]
    .groupby (gid [miss ])
    .agg (start =("timestamp_local","min"),
    end =("timestamp_local","max"),
    n_rows =("timestamp_local","size")))
    return runs
```

```python
import numpy as np
import pandas as pd
from sklearn .linear_model import LinearRegression
from sklearn .ensemble import HistGradientBoostingRegressor
from sklearn .metrics import mean_absolute_error ,mean_squared_error

d4 =(pd .read_csv ("system_4.csv",parse_dates =["timestamp_local"])
.sort_values ("timestamp_local").reset_index (drop =True ))

CAPACITY_W =1000
TARGET ="ac_power_w"
FEATURES =["irradiance","module_temp","ambient_temp"]

df =d4 [d4 ["irradiance"]>=150 ].copy ()
n_day =len (df )
df =df .dropna (subset =[TARGET ]+FEATURES )
n_complete =len (df )

train =df [df ["timestamp_local"]<="2018-08-03 23:59:59"].copy ()
test =df [df ["timestamp_local"]>="2019-03-20"].copy ()

k0 =LinearRegression (fit_intercept =False ).fit (
train [["irradiance"]],train [TARGET ]).coef_ [0 ]
healthy =train [TARGET ]>0.1 *k0 *train ["irradiance"]

not_over =train [TARGET ]<=CAPACITY_W

n_before =len (train )
train =train [healthy &not_over ]
print (f"Daylight: {n_day} | After excluding missing: {n_complete}")
print (f"Train: {n_before} -> {len(train)} after excluding zeros and over-capacity readings")
print (f"Test: {len(test)} (filtered only by daylight and missing values)")

base =LinearRegression (fit_intercept =False ).fit (train [["irradiance"]],train [TARGET ])

gbr =HistGradientBoostingRegressor (max_iter =300 ,learning_rate =0.05 ,
random_state =42 )
gbr .fit (train [FEATURES ],train [TARGET ])

def metrics (y ,pred ):
    mae =mean_absolute_error (y ,pred )
    rmse =np .sqrt (mean_squared_error (y ,pred ))
    return mae ,rmse ,100 *mae /CAPACITY_W ,100 *rmse /CAPACITY_W

rows =[]
for name ,mdl ,cols in [("Baseline (k*G)",base ,["irradiance"]),
("HistGradientBoosting",gbr ,FEATURES )]:
    for split ,part in [("train",train ),("test",test )]:
        mae ,rmse ,mae_pct ,rmse_pct =metrics (part [TARGET ],mdl .predict (part [cols ]))
        rows .append ([name ,split ,mae ,rmse ,mae_pct ,rmse_pct ])

res =pd .DataFrame (rows ,columns =["model","split","MAE_W","RMSE_W",
"MAE_%cap","RMSE_%cap"]).round (2 )
print (res .to_string (index =False ))
```

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib .pyplot as plt

from sklearn .ensemble import HistGradientBoostingRegressor
from sklearn .metrics import mean_absolute_error ,mean_squared_error ,r2_score

SYSTEM_ID =4
DAYLIGHT_IRRADIANCE =50
TEST_SIZE =0.20

DATA_FOLDER = Path(".")

TARGET ="ac_power_w"

FEATURES =[
"irradiance",
"module_temp",
"ambient_temp"
]

df =pd .read_csv (DATA_FOLDER /f"system_{SYSTEM_ID}.csv")

metadata =pd .read_csv (DATA_FOLDER /"systems_metadata.csv")

capacity_kw =metadata .loc [
metadata ["system_id"]==SYSTEM_ID ,
"capacity_kw"
].iloc [0 ]

capacity_w =capacity_kw *1000

df ["timestamp_local"]=pd .to_datetime (df ["timestamp_local"])

df =df .sort_values ("timestamp_local").reset_index (drop =True )

print ("Row count before cleaning:",len (df ))
print ("Nominal system capacity (W):",capacity_w )

df =df .dropna ().copy ()

print ("Row count after dropping missing values:",len (df ))

training_data =df .loc [

(df ["irradiance"]>=DAYLIGHT_IRRADIANCE )

&(~df ["is_night_offset"])

&(~df ["is_outlier_power"])

&(~df ["is_over_capacity"])
&(df ["ac_power_w"]<=capacity_w )

&(~df ["is_daylight_missing_power"])
&(df ["ac_power_w"]>0 )
].copy ()

print ("Daylight readings count suitable for training:",len (training_data ))

split_time =training_data ["timestamp_local"].quantile (1 -TEST_SIZE )

train_df =training_data .loc [
training_data ["timestamp_local"]<split_time
].copy ()

test_df =training_data .loc [
training_data ["timestamp_local"]>=split_time
].copy ()

print ("\nTraining period end:",train_df ["timestamp_local"].max ())
print ("Test period start:",test_df ["timestamp_local"].min ())

print ("Training rows count:",len (train_df ))
print ("Test rows count:",len (test_df ))

X_train =train_df [FEATURES ]
y_train =train_df [TARGET ]

X_test =test_df [FEATURES ]
y_test =test_df [TARGET ]

model =HistGradientBoostingRegressor (
learning_rate =0.08 ,
max_iter =200 ,
max_leaf_nodes =31 ,
l2_regularization =0.5 ,
early_stopping =False ,
random_state =42
)

model .fit (X_train ,y_train )

predicted_power_w =model .predict (X_test )

mae =mean_absolute_error (y_test ,predicted_power_w )
rmse =np .sqrt (mean_squared_error (y_test ,predicted_power_w ))
r2 =r2_score (y_test ,predicted_power_w )

print ("\nModel performance on recent test period:")
print (f"MAE  : {mae:.2f} W")
print (f"RMSE : {rmse:.2f} W")
print (f"R²   : {r2:.3f}")

results =test_df [
[
"timestamp_local",
"irradiance",
"module_temp",
"ambient_temp",
"ac_power_w"
]
].copy ()

results ["expected_power_w"]=predicted_power_w

results ["shortfall_w"]=(
results ["expected_power_w"]-results ["ac_power_w"]
)

results ["relative_shortfall"]=(
results ["shortfall_w"]/results ["expected_power_w"]
)

print ("\nFirst 10 rows of predictions:")
print (results .head (10 ).to_string (index =False ))

plot_data =results .head (2000 )

plt .figure (figsize =(14 ,6 ))

plt .plot (
plot_data ["timestamp_local"],
plot_data ["ac_power_w"],
label ="Actual power",
linewidth =1
)

plt .plot (
plot_data ["timestamp_local"],
plot_data ["expected_power_w"],
label ="Expected power",
linewidth =1.5
)

plt .xlabel ("Time")
plt .ylabel ("AC power (W)")
plt .title (f"Actual vs Expected AC Power — System {SYSTEM_ID}")
plt .legend ()
plt .grid (alpha =0.3 )
plt .tight_layout ()
plt .show ()
```

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib .pyplot as plt

from sklearn .ensemble import HistGradientBoostingRegressor
from sklearn .metrics import mean_absolute_error ,mean_squared_error ,r2_score

SYSTEM_ID =4

DATA_FOLDER = Path(".")

FEATURES =[
"irradiance",
"module_temp",
"ambient_temp"
]

TARGET ="ac_power_w"

MIN_IRRADIANCE_FOR_MODEL =50

MIN_IRRADIANCE_FOR_EVENT =150

MIN_EXPECTED_POWER_SHARE =0.20

MIN_EVENT_DURATION_MINUTES =30

CADENCE_MINUTES ={
4 :1 ,
10 :1 ,
1433 :15
}

cadence_minutes =CADENCE_MINUTES [SYSTEM_ID ]
cadence_hours =cadence_minutes /60

df =pd .read_csv (DATA_FOLDER /f"system_{SYSTEM_ID}.csv")
metadata =pd .read_csv (DATA_FOLDER /"systems_metadata.csv")

capacity_kw =metadata .loc [
metadata ["system_id"]==SYSTEM_ID ,
"capacity_kw"
].iloc [0 ]

capacity_w =float (capacity_kw *1000 )

df ["timestamp_local"]=pd .to_datetime (df ["timestamp_local"])
df =df .sort_values ("timestamp_local").reset_index (drop =True )

print ("Row count before cleaning:",len (df ))
print ("Nominal capacity (W):",capacity_w )

df =df .dropna ().copy ()

print ("Row count after dropping missing values:",len (df ))

reference_data =df .loc [
(df ["irradiance"]>=MIN_IRRADIANCE_FOR_MODEL )
&(~df ["is_night_offset"])
&(~df ["is_outlier_power"])
&(~df ["is_over_capacity"])
&(~df ["is_daylight_missing_power"])
&(df ["ac_power_w"]>0 )
&(df ["ac_power_w"]<=capacity_w )
].copy ()

print ("Clean reference readings count:",len (reference_data ))

train_cutoff =reference_data ["timestamp_local"].quantile (0.60 )
calibration_cutoff =reference_data ["timestamp_local"].quantile (0.80 )

train_df =reference_data .loc [
reference_data ["timestamp_local"]<train_cutoff
].copy ()

calibration_df =reference_data .loc [
(reference_data ["timestamp_local"]>=train_cutoff )
&(reference_data ["timestamp_local"]<calibration_cutoff )
].copy ()

test_df =reference_data .loc [
reference_data ["timestamp_local"]>=calibration_cutoff
].copy ()

print ("\nTraining period:")
print (train_df ["timestamp_local"].min (),"to",train_df ["timestamp_local"].max ())

print ("\nCalibration period:")
print (
calibration_df ["timestamp_local"].min (),
"to",
calibration_df ["timestamp_local"].max ()
)

print ("\nEvent detection & test period:")
print (test_df ["timestamp_local"].min (),"to",test_df ["timestamp_local"].max ())

print ("\nTraining rows count:",len (train_df ))
print ("Calibration rows count:",len (calibration_df ))
print ("Test rows count:",len (test_df ))

X_train =train_df [FEATURES ]
y_train =train_df [TARGET ]

model =HistGradientBoostingRegressor (
learning_rate =0.08 ,
max_iter =200 ,
max_leaf_nodes =31 ,
l2_regularization =0.5 ,
early_stopping =False ,
random_state =42
)

model .fit (X_train ,y_train )

def add_predictions (data ,trained_model ):
    output =data .copy ()

    output ["expected_power_w"]=trained_model .predict (
    output [FEATURES ]
    )

    output ["shortfall_w"]=(
    output ["expected_power_w"]-output ["ac_power_w"]
    )

    output ["relative_shortfall"]=np .where (
    output ["expected_power_w"]>0 ,
    output ["shortfall_w"]/output ["expected_power_w"],
    np .nan
    )

    return output

calibration_results =add_predictions (calibration_df ,model )

min_expected_power_w =capacity_w *MIN_EXPECTED_POWER_SHARE

calibration_for_threshold =calibration_results .loc [
(calibration_results ["irradiance"]>=MIN_IRRADIANCE_FOR_EVENT )
&(calibration_results ["expected_power_w"]>=min_expected_power_w )
].copy ()

shortfall_threshold_w =calibration_for_threshold [
"shortfall_w"
].quantile (0.95 )

relative_shortfall_threshold =calibration_for_threshold [
"relative_shortfall"
].quantile (0.95 )

print ("\nShortfall threshold (W) from calibration data:")
print (round (shortfall_threshold_w ,2 ),"W")

print ("\nRelative shortfall threshold from calibration data:")
print (round (relative_shortfall_threshold *100 ,2 ),"%")

print ("\nMinimum expected power threshold for ratio calculation:")
print (round (min_expected_power_w ,2 ),"W")

test_results =add_predictions (test_df ,model )

mae =mean_absolute_error (
test_results ["ac_power_w"],
test_results ["expected_power_w"]
)

rmse =np .sqrt (
mean_squared_error (
test_results ["ac_power_w"],
test_results ["expected_power_w"]
)
)

r2 =r2_score (
test_results ["ac_power_w"],
test_results ["expected_power_w"]
)

print ("\nModel performance on final test period:")
print (f"MAE:  {mae:.2f} W")
print (f"RMSE: {rmse:.2f} W")
print (f"R²:   {r2:.3f}")

scoring_data =df .loc [
(df ["timestamp_local"]>=calibration_cutoff )
&(df ["irradiance"]>=MIN_IRRADIANCE_FOR_EVENT )
&(~df ["is_night_offset"])
&(~df ["is_outlier_power"])
&(~df ["is_over_capacity"])
].copy ()

scoring_results =add_predictions (scoring_data ,model )

scoring_results ["candidate_shortfall"]=(
(scoring_results ["expected_power_w"]>=min_expected_power_w )
&(scoring_results ["shortfall_w"]>shortfall_threshold_w )
&(
scoring_results ["relative_shortfall"]
>relative_shortfall_threshold
)
)

candidate_rows =scoring_results .loc [
scoring_results ["candidate_shortfall"]
].copy ()

candidate_rows =candidate_rows .sort_values (
"timestamp_local"
).reset_index (drop =True )

print ("\nCandidate readings count for inspection:",len (candidate_rows ))

print (
"including zero or negative power during daylight:",
candidate_rows ["is_daylight_missing_power"].sum ()
)

candidate_rows ["minutes_since_previous"]=(
candidate_rows ["timestamp_local"]
.diff ()
.dt .total_seconds ()
.div (60 )
)

candidate_rows ["new_event"]=(
candidate_rows ["minutes_since_previous"].isna ()
|(
candidate_rows ["minutes_since_previous"]
>cadence_minutes
)
)

candidate_rows ["event_id"]=(
candidate_rows ["new_event"].cumsum ()
)

candidate_rows ["expected_energy_part_kwh"]=(
candidate_rows ["expected_power_w"]
*cadence_hours
/1000
)

candidate_rows ["actual_energy_part_kwh"]=(
candidate_rows ["ac_power_w"]
*cadence_hours
/1000
)

candidate_rows ["lost_energy_part_kwh"]=(
candidate_rows ["shortfall_w"]
.clip (lower =0 )
*cadence_hours
/1000
)

events =(
candidate_rows
.groupby ("event_id",as_index =False )
.agg (
system_id =("system_id","first"),
start_time =("timestamp_local","min"),
end_time =("timestamp_local","max"),
qualifying_readings =("timestamp_local","size"),
average_irradiance =("irradiance","mean"),
average_actual_power_w =("ac_power_w","mean"),
average_expected_power_w =("expected_power_w","mean"),
average_shortfall_w =("shortfall_w","mean"),
peak_shortfall_w =("shortfall_w","max"),
peak_relative_shortfall =(
"relative_shortfall",
"max"
),
daylight_missing_power_readings =(
"is_daylight_missing_power",
"sum"
),
expected_energy_kwh =(
"expected_energy_part_kwh",
"sum"
),
actual_energy_kwh =(
"actual_energy_part_kwh",
"sum"
),
estimated_lost_energy_kwh =(
"lost_energy_part_kwh",
"sum"
)
)
)

events ["duration_minutes"]=(
events ["qualifying_readings"]*cadence_minutes
)

events ["duration_hours"]=(
events ["duration_minutes"]/60
)

events =events .loc [
events ["duration_minutes"]>=MIN_EVENT_DURATION_MINUTES
].copy ()

events =events .sort_values (
"estimated_lost_energy_kwh",
ascending =False
).reset_index (drop =True )

print ("\nEvents count sustained for >= 30 minutes:")
print (len (events ))

display_columns =[
"event_id",
"start_time",
"end_time",
"duration_minutes",
"average_irradiance",
"average_shortfall_w",
"peak_shortfall_w",
"estimated_lost_energy_kwh",
"daylight_missing_power_readings"
]

print ("\nTop 20 events by lost energy:")
print (
events [display_columns ]
.head (20 )
.to_string (index =False )
)

if len (events )>0 :
    top_event =events .iloc [0 ]

    plot_start =(
    top_event ["start_time"]
    -pd .Timedelta (minutes =30 )
    )

    plot_end =(
    top_event ["end_time"]
    +pd .Timedelta (minutes =30 )
    )

    event_plot =scoring_results .loc [
    (scoring_results ["timestamp_local"]>=plot_start )
    &(scoring_results ["timestamp_local"]<=plot_end )
    ].copy ()

    event_candidates =event_plot .loc [
    event_plot ["candidate_shortfall"]
    ].copy ()

    plt .figure (figsize =(14 ,6 ))

    plt .plot (
    event_plot ["timestamp_local"],
    event_plot ["ac_power_w"],
    label ="Actual power",
    linewidth =1.5
    )

    plt .plot (
    event_plot ["timestamp_local"],
    event_plot ["expected_power_w"],
    label ="Expected power",
    linewidth =1.5
    )

    plt .scatter (
    event_candidates ["timestamp_local"],
    event_candidates ["ac_power_w"],
    color ="red",
    s =20 ,
    label ="Candidate readings",
    zorder =3
    )

    plt .xlabel ("Time")
    plt .ylabel ("AC power (W)")
    plt .title (
    f"Top candidate event — System {SYSTEM_ID}"
    )
    plt .legend ()
    plt .grid (alpha =0.3 )
    plt .tight_layout ()
    plt .show ()
```

```python
def add_model_results (data ,trained_model ):
    result =data .copy ()

    result ["expected_power_w"]=trained_model .predict (
    result [FEATURES ]
    )

    result ["shortfall_w"]=(
    result ["expected_power_w"]
    -result ["ac_power_w"]
    )

    result ["residual_w"]=(
    result ["ac_power_w"]
    -result ["expected_power_w"]
    )

    result ["performance_ratio"]=np .where (
    result ["expected_power_w"]>0 ,
    result ["ac_power_w"]
    /result ["expected_power_w"],
    np .nan
    )

    return result
```

```python
import numpy as np
import pandas as pd

FLAG_COLUMNS =["is_night_offset","is_outlier_power",
"is_over_capacity","is_daylight_missing_power"]
INJECT_START_HOUR =10
INJECT_HOURS =3
N_INJECTIONS =30
FAULT_LEVELS ={"drop_10pct":0.9 ,"drop_20pct":0.8 ,"zero_power":0.0 }
RANDOM_SEED =42

def get_time (d ):

    col =d ["timestamp_local"]if "timestamp_local"in d .columns else d .index .to_series ()
    return pd .to_datetime (col )

def detect_events (d ,power ,expected ,thr_w ,thr_rel ,min_exp ,cadence_min ):
    ts =get_time (d )
    flagged =d [FLAG_COLUMNS ].fillna (False ).astype (bool ).any (axis =1 )
    shortfall =expected -power
    relative =shortfall /expected
    ok =(d ["irradiance"]>=MIN_IRRADIANCE_FOR_EVENT )&(~flagged )&(expected >=min_exp )
    candidate =ok &(shortfall >thr_w )&(relative >thr_rel )

    t =ts [candidate ]
    if len (t )==0 :
        return pd .DataFrame (columns =["start","end","duration_min"])
    group_id =(t .diff ()>pd .Timedelta (minutes =cadence_min )).cumsum ()
    ev =t .groupby (group_id ).agg (start ="min",end ="max").reset_index (drop =True )
    ev ["duration_min"]=(ev ["end"]-ev ["start"]).dt .total_seconds ()/60 +cadence_min
    return ev [ev ["duration_min"]>=MIN_EVENT_DURATION_MINUTES ].reset_index (drop =True )
```

```python
def fault_injection_test (d ,model ,features ,thr_w ,thr_rel ,min_exp ,
cadence_min ,system_id ,healthy_end =None ):
    d =d .copy ()
    ts =get_time (d )
    assert ts .is_monotonic_increasing ,"Data must be sorted by timestamp"

    if healthy_end is not None :
        keep =ts <=pd .Timestamp (healthy_end )
        d ,ts =d [keep ],ts [keep ]

    X =d [features ]
    valid_x =X .notna ().all (axis =1 )
    expected =pd .Series (np .nan ,index =d .index )
    expected [valid_x ]=model .predict (X [valid_x ])
    power =d [TARGET ]

    base_events =detect_events (d ,power ,expected ,thr_w ,thr_rel ,min_exp ,cadence_min )
    base_days =set (base_events ["start"].dt .normalize ())if len (base_events )else set ()

    day =ts .dt .normalize ()
    hour =ts .dt .hour +ts .dt .minute /60
    in_window =(hour >=INJECT_START_HOUR )&(hour <INJECT_START_HOUR +INJECT_HOURS )
    flagged =d [FLAG_COLUMNS ].fillna (False ).astype (bool ).any (axis =1 )
    ok_rows =((d ["irradiance"]>=MIN_IRRADIANCE_FOR_EVENT )&(~flagged )
    &(expected >=min_exp )&power .notna ())
    valid_count =(in_window &ok_rows ).groupby (day ).sum ()
    rows_needed =0.9 *INJECT_HOURS *60 /cadence_min
    good_days =[x for x in valid_count .index [valid_count >=rows_needed ]if x not in base_days ]

    n =min (N_INJECTIONS ,len (good_days ))
    rng =np .random .default_rng (RANDOM_SEED )
    chosen =pd .to_datetime (rng .choice (np .array (good_days ),size =n ,replace =False ))
    chosen =sorted (chosen )
    windows =[(x +pd .Timedelta (hours =INJECT_START_HOUR ),
    x +pd .Timedelta (hours =INJECT_START_HOUR +INJECT_HOURS ))for x in chosen ]
    inject_mask =in_window &day .isin (chosen )

    rows =[]
    for name ,factor in FAULT_LEVELS .items ():
        p_inj =power .copy ()
        p_inj [inject_mask ]=power [inject_mask ]*factor
        ev =detect_events (d ,p_inj ,expected ,thr_w ,thr_rel ,min_exp ,cadence_min )

        detected =sum (((ev ["start"]<w1 )&(ev ["end"]>=w0 )).any ()for w0 ,w1 in windows )
        outside =sum (not any ((s <w1 )and (e >=w0 )for w0 ,w1 in windows )
        for s ,e in zip (ev ["start"],ev ["end"]))
        rows .append ({"system":system_id ,"fault":name ,"injected_days":n ,
        "detected":detected ,"recall %":round (100 *detected /n ,1 ),
        "events_outside":outside })

    print (f"System {system_id}: valid days = {len(good_days)}, events before injection = {len(base_events)}")
    return pd .DataFrame (rows )
```
